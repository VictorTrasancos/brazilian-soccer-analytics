```python
import pandas as pd
import numpy as np
from sqlalchemy import create_engine

# 1. Carrega o CSV original
df = pd.read_csv('../data/brasileirao_matches.csv')

# 2. Executa a limpeza básica (remove nulos e padroniza)
df_limpo = df.dropna().copy()
df_limpo = df_limpo.rename(columns={
    'datetime': 'data_jogo',
    'home_team': 'mandante',
    'away_team': 'visitante',
    'home_team_state': 'estado_mandante',
    'away_team_state': 'estado_visitante',
    'home_goal': 'gols_mandante',
    'away_goal': 'gols_visitante'
})

# Ajusta a data para o formato limpo
df_limpo['data_jogo'] = pd.to_datetime(df_limpo['data_jogo']).dt.date
df_limpo['mandante'] = df_limpo['mandante'].str.strip()
df_limpo['visitante'] = df_limpo['visitante'].str.strip()
df_limpo = df_limpo.drop(columns=['season', 'round'], errors='ignore')

```

```python
import urllib.parse
from sqlalchemy import create_engine
import pandas as pd

usuario = 'root'
senha_original = 'V//03fff'  

# Codifica a senha de forma segura para URLs (trata as barras /)
senha_segura = urllib.parse.quote_plus(senha_original)

host = 'localhost'
porta = '3306'
nome_banco = 'brasileirao_db'

# Monta a URL com a senha tratada
url_conexao = f'mysql+pymysql://{usuario}:{senha_segura}@{host}:{porta}/{nome_banco}'
engine = create_engine(url_conexao)

try:
    # Envia o df_limpo para virar a tabela 'partidas' no MySQL
    df_limpo.to_sql(name='partidas', con=engine, if_exists='replace', index=False)
    print("Sucesso absoluto! Dados enviados para o MySQL com sucesso.")
except Exception as e:
    print(f"Erro na conexão: {e}")
```
