```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Carregando o dataset
df = pd.read_csv('../data/brasileirao_matches.csv')

display(df.head(10))

print('Informações técnias do dataset')
df.info()

print('Valores ausentes por coluna')
print(df.isna().sum())
```

```python
df_limpo = df.dropna().copy()
df_limpo = df.drop_duplicates()

print('Valores nulos restantes')
print(df_limpo.isna().sum())
```

```python
df_limpo['datetime'] = pd.to_datetime(df_limpo['datetime'])
df_limpo['home_team'] = df_limpo['home_team'].str.strip()
df_limpo['away_team'] = df_limpo['away_team'].str.strip()
df_limpo = df_limpo.drop(columns=['season', 'round'])
```

```python
df_limpo = df_limpo.rename(columns={'datetime': 'data_jogo'})
df_limpo = df_limpo.rename(columns={'home_team':'mandante'})
df_limpo = df_limpo.rename(columns={'away_team':'visitante'})
df_limpo = df_limpo.rename(columns={'home_team_state':'estado_mandante'})
df_limpo = df_limpo.rename(columns={'away_team_state':'estado_visitante'})
df_limpo = df_limpo.rename(columns={'home_goal':'gols_mandante'})
df_limpo = df_limpo.rename(columns={'away_goal':'gols_visitante'})
```

```python
df_limpo['data_jogo'] = pd.to_datetime(df_limpo['data_jogo'])
df_limpo['mandante'] = df_limpo['mandante'].str.strip()
df_limpo['visitante'] = df_limpo['visitante'].str.strip()
df_limpo = df_limpo.drop(columns=['season', 'round'], errors='ignore')
```
